Q2) create-dataframe

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark=(
    SparkSession.builder.appName("PysparkInstallationTest")
    .master("local[*]").getOrCreate()
)
cric_path="/content/cricket_players.csv"
cricket_df=spark.read.csv(cric_path,header=True,inferSchema=True)
cricket_df.show()
```

```python
print(cricket_df.count())
```

```python
cricket_df.printSchema()
```

```python
df1=cricket_df.select("player_name","country").orderBy("country",ascending=False)
#(or)
#df1=cricket_df.select("player_name","country").orderBy(col("country").desc())
df1.show()
```

```python
cricket_df.select("player_name","country","age").orderBy(["country", "age"],ascending=[1, 0]).show()
```

```python
cricket_df.select("player_name","country","runs").filter((col("country")=="India") & (col("runs")<1000)).show()
# (or)
# cricket_df.select("player_name","country","runs").filter("country='India' and runs<1000").show()
```

```python
cricket_df.groupBy("country").count().show()
```

```python
#where
cricket_df.select("country","runs").filter("country='India'").groupby("country").agg(sum("runs").alias("Total_runs")).show()
```

```python
#having
cricket_df.select("country","runs").filter("country='India' or country='Bangladesh'").groupby("country").agg(sum("runs").alias("Total_runs")).filter(col("Total_runs")>10000).show()
# (or)
#having
# cricket_df.select("country","runs").filter("country in ('India','Bangladesh')").groupby("country").agg(sum("runs").alias("Total_runs")).filter(col("Total_runs")>10000).show()
```
